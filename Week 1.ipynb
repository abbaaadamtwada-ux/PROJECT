```python
#Write a program that calculates the area of a rectangle.

#Your program should:

#Ask the user for the length
#Ask the user for the width
#Calculate the area (length × width)
#Calculate the perimeter (2 × length + 2 × width)
#Print both results with descriptive messages

length = float(input("Enter the length: "))
width = float(input("Enter the width: "))
area = length * width
perimeter = 2 * length + 2 * width
print(f"The area of the rectangle is: {area}")
print(f"The perimeter of the rectangle is: {perimeter}")
```

```python
#Create a program that calculates a shopping bill.

#Your program should:

#Ask for the price of three items
#Calculate the subtotal (sum of all prices)
#Calculate the sales tax (6% of subtotal)
#Calculate the final total (subtotal + tax)
#Print the subtotal, tax, and final total

price1 = float(input("Enter the price item 1: "))
price2 = float(input("Enter the price item 2: "))
price3 = float(input("Enter the price item 3: "))
subtotal = price1 + price2 + price3
tax = subtotal * 0.06
total = subtotal + tax
print(f"Subtotal: ${subtotal:.2f}")
print(f"Sales Tax (6%): ${tax:.2f}")
print(f"Final Total: ${total:.2f}")
```
