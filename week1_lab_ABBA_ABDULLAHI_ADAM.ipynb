```python
# EXERCISE 3.FIzzBuzz
output = []
for n in range(1, 31):
    if n % 3 == 0 and n % 5 == 0:
        output.append('FizzBuzz')
    elif n % 3 == 0:
        output.append("Fizz")
    elif n % 5 == 0:
        output.append("Buzz")
    else:
        output.append(n)
    print(output)
```

```python
# CHECK
_expected = ["FizzBuzz" if k % 15 == 0 else "Fizz" if k % 3 == 0 else "Buzz" if k % 5 == 0 else k for k in range(1, 31)]
assert len(output) == 30, "output should have 30 entries, you have " + str(len(output))
for _k in range(30):
    assert output[_k] == _expected[_k], "entry for the number " + str(_k + 1) + " should be " + repr(_expected[_k]) + ", you got " + repr(output[_k])
print("Looks good.")
```

```python
# EXERCISE GRADE STATISTIC
scores = [72, 45, 88, 50, 91, 33, 67, 49, 100, 58]

highest = scores[0]
lowest = scores[0]
total = 0
passCount = 0
count = 1

for score in scores:
    if score > highest:
        highest = score
    if score < lowest:
        lowest = score
    total += 1
    if score >= 50:
        passCount += 1

average = total / count

print(highest, lowest, average, passCount)
```

```python
# CHECK
highest == 100, "highest should be 100, you got " + repr(highest)
lowest == 33, "lowest should be 33, you got " + repr(lowest)
abs(average - 65.3) < 1e-9, "average should be 65.3, you got " + repr(average)
passCount == 7, "7 scores are 50 or above (remember that 50 itself is a pass), you got " + repr(passCount)
print("Looks good.")
```

```python
import random

def guessing_game():
    """Play a number guessing game.

    The computer pick a random number between 1 and 100.
    The player gets at most 7 valid guesses.
    Inputs outside 1-100 are rejected
    """

    secret = random.randint(1, 100)

    guesses_used = 0
    max_guesses = 7

    while guesses_used < max_guesses:
        try:
            guess = int(input(f"Guess {guesses_used + 1}/{max_guesses} (1-100): "))
        except ValueError:
            print("Please enter a whole number.")
            continue

        if guess < 1 or guess > 100:
            print("Out of range! Enter a number between 1 and 100.")
            continue

        guesses_used += 1

        if guess == secret:
            print(f"Correct! You won in {guesses_used} guesses.")
            return
        elif guess < secret:
            print("Too small")
        else:
            print("Too large")

    print(f"Out of guesses! The number was {secret}.")

guessing_game()
```
