# ANSWER TO THE QUESTION
The condition 1 is satisfied that is why it gives no chance to the remaining condition, and if we need the below condition to work then we are to use lessthan or equal to then it will give chance to other condition all it must follow sequencially if other conditions want be met.

```python


hasIdCard = True
hasPaidFees = False

# and: BOTH must be True
print("May enter the hall:", hasIdCard and hasPaidFees)

# or: ONE is enough
hasCash = False
hasTransferApp = True
print("Can pay:", hasCash or hasTransferApp)

# not: flips True <-> False
isAbsent = False
print("Is present:", not isAbsent)

# The range check: is the age between 18 and 35, both included?
age = 27
isYouth = age >= 18 and age <= 35
print("Qualifies for the youth programme:", isYouth)

# The opposite question: is the age OUTSIDE that range?  Two ways, same answer.
print(age < 18 or age > 35)
print(not (age >= 18 and age <= 35))

# A compound condition inside an if:
score = 82
attendance = 60
if score >= 50 and attendance >= 75:
    print("Certificate awarded.")
else:
    print("No certificate: you need a pass mark AND 75 percent attendance.")
```
